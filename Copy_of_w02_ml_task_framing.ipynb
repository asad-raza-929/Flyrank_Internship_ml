## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric I would defend for this churn classification task is the **F1-score**.

**Why F1-score?**

*   **Imbalanced Classes:** Churn datasets are often imbalanced, meaning the number of non-churning customers (negative class) significantly outweighs the number of churning customers (positive class). Metrics like accuracy can be misleading in such cases, as a model that always predicts 'no churn' might achieve high accuracy simply by predicting the majority class correctly, while failing to identify actual churners.
*   **Balancing Precision and Recall:**
    *   **Precision** (of the positive class, i.e., churn): Of all customers predicted to churn, how many actually churned? High precision means fewer false positives (contacting customers who weren't actually going to churn).
    *   **Recall** (of the positive class, i.e., churn): Of all customers who *actually* churned, how many did we correctly predict? High recall means fewer false negatives (missing actual churners).
    *   The F1-score is the harmonic mean of precision and recall, providing a single metric that balances both. This is crucial for churn prediction because both false positives (wasted retention effort) and false negatives (lost customers) have significant business costs.

**What number means 'good'?**

A "good" F1-score is context-dependent, but generally, an F1-score **above 0.7 (70%)** would be considered a strong performance, indicating a good balance between identifying churners and avoiding misclassifications. For a critical business problem like churn, we would aim for an even higher score, perhaps **0.8 or greater**, depending on the baseline and complexity of the data. A comparison against a simple baseline (e.g., predicting everyone will not churn) would also be essential to establish what constitutes a meaningful improvement.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Machine learning beats a fixed rule for predicting customer churn because the underlying patterns are typically **complex, non-linear, and involve interactions between many variables**, making them too messy to capture accurately with simple `if-else` statements.

Here's why ML is superior:

1.  **Multidimensionality and Interactions:** Churn is rarely caused by a single factor. It's often a combination of factors like high monthly spend *combined with* low feature usage, or a recent support ticket *followed by* a period of inactivity, but *only* for customers with a short `tenure_months`. A fixed rule would require an explosion of `if-elif-else` conditions to cover all these interactions and thresholds.
2.  **Non-linear Relationships:** The relationship between features and churn isn't always linear. For example, a customer with very low `monthly_spend` might be at risk, but so might a customer with unusually high spend (indicating dissatisfaction or a one-time purchase). A simple threshold (`if monthly_spend < X then churn`) wouldn't capture this U-shaped or other complex relationships.
3.  **Dynamic Patterns:** The factors influencing churn can change over time (e.g., competitive landscape, new product features, economic conditions). Fixed rules are static and would need constant manual updates, whereas ML models can be retrained periodically to adapt to evolving patterns.
4.  **Feature Importance and Weighting:** Not all features contribute equally to churn, and their importance can vary. ML models automatically learn the relative importance and weights of different features, giving more predictive power to relevant indicators without needing explicit human-defined rules for each.
5.  **Probabilistic Outputs:** ML models can provide a probability of churn, not just a binary yes/no. This allows for nuanced decision-making, such as targeting retention efforts based on the level of risk, which a fixed rule cannot provide.

In essence, the 'messiness' comes from the subtle, interconnected, and evolving nature of human behavior and business dynamics, which ML algorithms are specifically designed to model.
