## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

The unit of analysis for this data contract is **daily search query performance per country and device type**. Each row represents the aggregated performance metrics (e.g., clicks, impressions, conversions) for a unique combination of `date`, `query`, `country`, and `device_type`.

The time window covers the **last 30 days** of data, providing recent trends and performance insights.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
import pandas as pd
import numpy as np
from datetime import date, timedelta

# Generate synthetic data for demonstration

# Define the time window (last 30 days)
end_date = date.today()
start_date = end_date - timedelta(days=29)
dates = pd.date_range(start=start_date, end=end_date, freq='D')

queries = ['how to make coffee', 'best local restaurants', 'weather forecast', 'buy new phone', 'learn python']
countries = ['USA', 'UK', 'Germany']
device_types = ['desktop', 'mobile', 'tablet']

data = []
for dt in dates:
    for query in queries:
        for country in countries:
            for device in device_types:
                impressions = np.random.randint(100, 10000)
                clicks = np.random.randint(5, int(impressions * 0.1))
                # Ensure the 'high' argument for randint is always greater than 'low' (0)
                conversions = np.random.randint(0, max(1, int(clicks * 0.05)))
                data.append([dt, query, country, device, impressions, clicks, conversions])

df = pd.DataFrame(data, columns=['date', 'query', 'country', 'device_type', 'impressions', 'clicks', 'conversions'])

# Calculate CTR and Conversion Rate
df['ctr'] = (df['clicks'] / df['impressions']).fillna(0)
df['conversion_rate'] = (df['conversions'] / df['clicks']).fillna(0)

print("Sample of the generated data:")
display(df.head())

print(f"\nData covers dates from {df['date'].min().strftime('%Y-%m-%d')} to {df['date'].max().strftime('%Y-%m-%d')}")

# Verify the grain: check for duplicates based on the defined unit of analysis
duplicate_rows = df.duplicated(subset=['date', 'query', 'country', 'device_type']).sum()
print(f"\nNumber of duplicate rows for the defined grain: {duplicate_rows}")
if duplicate_rows == 0:
    print("Grain verified: Each row is unique for date, query, country, and device_type.")
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Understanding the limits of the data is crucial to avoid drawing incorrect conclusions or attempting to solve problems beyond its scope.

*   **Search-Only Perspective:** This data contract focuses solely on search engine performance. It does not provide information on user behavior *after* landing on a page (e.g., bounce rate, time on site, funnel progression beyond initial conversion), social media impact, or direct traffic.
*   **Aggregated Nature:** Because the data is aggregated daily by query, country, and device type, it's impossible to analyze individual user journeys or understand personalized search experiences.
*   **Limited Historical Depth:** While covering the last 30 days is useful for recent trends, it cannot provide insights into long-term seasonality, year-over-year growth, or performance during specific past events more than 30 days ago.
*   **Sampling Bias (Potential):** If the underlying data source is sampled (e.g., Google Search Console's data might be sampled for very high-volume queries), the reported numbers might not represent the absolute true volume, but rather a reliable directional indicator.
*   **No Competitive Intelligence:** The data reflects only *our* search performance and offers no direct insight into competitors' strategies, performance, or market share.
*   **Attribution Model:** The 'conversions' metric is as reported by the source system (e.g., Google Analytics). The specific attribution model (e.g., last click, data-driven) is external to this data contract and could influence interpretation.
